# Decorator

```python
def make_validator(min_value):
    def validator(x):
        return x >= min_value
    return validator
```

```python
is_adult = make_validator(18)
```

```python
print(is_adult(15))
```

```python
print(is_adult(18))
```

```python
def timer(func):
    import time
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)  # تابع اصلی را اینجا صدا می‌زنیم
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

def slow_function():
    import time
    time.sleep(1)
    return "Done"

# بدون @ decorator، دستی این‌طور استفاده می‌کنیم:
wrapped = timer(slow_function)
result = wrapped()
print(result)
```

```python
@timer
def slow_function():
    import time
    time.sleep(1)
    return "Done"

result = slow_function()
print(result)
```

```python
def log_call(func):
    def wrapper(*args , **kwargs):

        print(f'Calling {func.__name__.title()} with args = {args} , kwargs = {kwargs}')

        result = func(*args , **kwargs)

        return result
    return wrapper

@log_call
def add(a,b):
    return a + b

@log_call
def multiply(a,b):
    return a * b

print(add(4,6))
print('--------')
print(multiply(8,8))
```
