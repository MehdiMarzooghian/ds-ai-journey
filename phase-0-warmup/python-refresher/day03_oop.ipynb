```python
from math import sqrt
```

```python
class Dataset:
    def __init__(self , name , data):

        self.name = name
        self.data = data

    def __repr__(self):
        return f"Dataset(name : {self.name} , size : {self.size()} , mean : {self.mean()})"

    def mean(self):
        mean  = sum(self.data) / len(self.data)
        return mean


    def size(self):
        size = len(self.data)
        return size


class NumericDataset(Dataset):
    def super__init__(self , name , data)
        def std(self):
            result = sqrt( sum((x - mean)^2 for x in data) / len(data) )
            return result

class TextDataset(Dataset):
    def avg_length(self):
        result = sum(self.data) / len(self.data)
        return result
```

```python
nd = NumericDataset("Prices", [10, 20, 30, 40, 50])
print(nd.std())

td = TextDataset("Names", ["Ali", "Maryam", "Reza"])
print(td.avg_length())
```
