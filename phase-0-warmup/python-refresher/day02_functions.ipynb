```python
def train_model(model_name , **kwargs):
    print(f'Training model : {model_name}')
    for key , value in kwargs.items():
        print(f'   {key} : {value}')
```

```python
train_model("RandomForest" , arash = 29)
```

```python
train_model("RandomForest", lr=0.01, epochs=100, max_depth=5)
```
